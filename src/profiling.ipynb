```python
import pandas as pd
from pathlib import Path
```

Reading Files 

```python
PROJECT_ROOT = Path.cwd().parent
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

customers = pd.read_csv(RAW_DATA_DIR / "cust_info.csv")
products = pd.read_csv(RAW_DATA_DIR / "prd_info.csv")
sales = pd.read_csv(RAW_DATA_DIR / "sales_details.csv")
category = pd.read_csv(RAW_DATA_DIR / "PX_CAT_G1V2.csv")
location = pd.read_csv(RAW_DATA_DIR / "LOC_A101.csv")
additional_customers = pd.read_csv(RAW_DATA_DIR / "CUST_AZ12.csv")
```

Missing values

```python
def missing_values(df):
    report =  pd.DataFrame({"Missing_count": df.isna().sum()})
    return report.sort_values("Missing_count",ascending = False)
```

```python
display(missing_values(customers))
display(missing_values(products))
display(missing_values(sales))
display(missing_values(category))
display(missing_values(location))
display(missing_values(additional_customers))
```
